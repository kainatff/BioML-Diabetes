```python
# Diabetes Model Comparison in Google Colab
# This code integrates CFA, GA, and MLP results into one analysis notebook

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sample placeholder data: Replace with actual collected results
results = [
    {"Model": "CFA", "Dataset": "PID", "Accuracy": 0.7619, "Kappa": 0.4598, "MAE": 0.2381, "Time": 32.40},
    {"Model": "CFA", "Dataset": "HFD", "Accuracy": 0.7950, "Kappa": 0.5295, "MAE": 0.2050, "Time": 38.12},
    {"Model": "GA", "Dataset": "PID", "Accuracy": 0.78, "Kappa": 0.40, "MAE": 0.24, "Time": 64},
    {"Model": "GA", "Dataset": "HFD", "Accuracy": 0.96, "Kappa": 0.92, "MAE": 0.03, "Time": 75},
    {"Model": "MLP", "Dataset": "PID", "Accuracy": 0.7403, "Kappa": 0.4160, "MAE": 0.2597, "Time": 20.8919},
    {"Model": "MLP", "Dataset": "HFD", "Accuracy": 0.8167, "Kappa": 0.5926, "MAE": 0.1833, "Time": 33.4594},
]

# Create a DataFrame
results_df = pd.DataFrame(results)

# --- Visualization Functions ---
def plot_metrics(df, metric):
    plt.figure(figsize=(10, 6))
    sns.barplot(data=df, x="Model", y=metric, hue="Dataset")
    plt.title(f"Comparison of {metric} across Models and Datasets")
    plt.ylim(0, 1 if metric != "Time" else df[metric].max() + 10)
    plt.ylabel(metric)
    plt.legend(loc='best')
    plt.tight_layout()
    plt.show()

# Plot accuracy, kappa, MAE, and execution time
for metric in ["Accuracy", "Kappa", "MAE", "Time"]:
    plot_metrics(results_df, metric)

# --- Generate Summary Table ---
print("\n\nPerformance Summary Table:")
print(results_df.pivot(index="Model", columns="Dataset", values=["Accuracy", "Kappa", "MAE", "Time"]))
```
